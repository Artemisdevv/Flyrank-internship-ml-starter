## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## The question

### Research question

Which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring, based on observable search, content, and engagement signals?

### Decision

The decision is which pages should enter a limited content-review queue first.

### Unit of analysis

The unit of analysis is one content page.

For the starter dataset, each row represents a deduplicated content item after the starter feature preparation step.

### Output

The intended output is a ranked review queue. Each page should receive a priority score and, where possible, a reason code explaining why it was prioritized.

### Action

A human reviewer could use the ranked queue to decide whether a page should be refreshed, expanded, protected, pruned, or simply monitored.

### Cost of a wrong recommendation

A false positive would send a page to review when it did not need attention, wasting limited reviewer time.

A false negative would leave a potentially important page out of the review queue, meaning an opportunity or problem could be missed.

The goal is therefore not to guarantee that an action will improve a page. The goal is to make the limited review process more evidence-based and better prioritized.
